```python
# Part 3: Matplotlib
df=pd.merge(order,details,on="Order ID")
```

```python
df
```

```python
x=df["Category"].value_counts()
x
```

```python
x.index
```

```python
x.values
```

```python
# Bar chart of Category Sales
plt.figure(figsize=(4,3))
plt.bar(x.index,x.values,color="b")
plt.xlabel("category")
plt.ylabel("sales")
plt.title("Category-wise Sales")
plt.show()
```

```python
x=df["State"].value_counts()
x
```

```python
x.index
```

```python
x.values
```

```python
# Bar Chart of State Sales
plt.figure(figsize=(21,20))
plt.bar(x.index,x.values,color="green")
plt.xlabel("state")
plt.ylabel("sales")
plt.title("State-wise Sales")
plt.show()
```

```python
# Pie Chart of Category Contribution
plt.figure(figsize=(12,10))
plt.pie(x.values,labels=x.index)
plt.title("category contribution",color="lime",fontsize=15)
plt.show()
```

```python
# Line Chart of Monthly Sales
plt.figure(figsize=(8,5))
plt.plot(x.index,x.values,color="green",marker="o",markersize=5,mfc="r",mec="r")
plt.xlabel("month")
plt.ylabel("amount")
plt.title("monthly sales")
plt.xticks(list(range(1,13)),["jan","feb","mar","apr","may","june","july","aug","sept","oct","nov","dec"])
plt.show()
```

```python
# Histogram of Amount
plt.figure(figsize=(7,5))
plt.hist(df["Amount"], bins=20,color="r")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Distribution of Sales Amount")
plt.show()
```

```python
# Histogram of Profit
plt.figure(figsize=(7,5))
plt.hist(df["Profit"], bins=20,color="lime")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.title("Distribution of Profit")
plt.show()
```

```python
# Scatter Plot (Amount vs Profit)
plt.figure(figsize=(7,5))
plt.scatter(x="Amount",y="Profit",data=df)
plt.xlabel("Amount")
plt.ylabel("Profit")
plt.title("Amount vs Profit")
plt.show()
```

```python
# Boxplot of Profit
plt.figure(figsize=(6,5))
plt.boxplot(df["Profit"],data=df)
plt.ylabel("Profit")
plt.title("Boxplot of Profit")
plt.show()
```
