```python
columns=["MultipleLines","OnlineSecurity","OnlineBackup","DeviceProtection","TechSupport","StreamingTV","StreamingMovies"]
```

```python
for i in columns:
    df[i]=df[i].apply(lambda i:"No" if i=="No internet service" else i)
```

```python
for i in df.columns:
    if df[i].dtype=="object":
        print(i)
        print(df[i].unique())
```

```python
df["MultipleLines"]=df["MultipleLines"].apply(lambda i:"No" if i=='No phone service' else i)
```

```python
for i in df.columns:
    if df[i].dtype=="object":
        print(i)
        print(df[i].unique())
```

```python
df["gender"]=df["gender"].map({"Female":0,"Male":1})
```

```python
for i in df.columns:
    if df[i].dtype=="object":
        print(i)
        print(df[i].unique())
```

```python
df.columns
```

```python
col=['Partner', 'Dependents',
       'PhoneService', 'MultipleLines', 'OnlineSecurity',
       'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV',
       'StreamingMovies','PaperlessBilling', 'Churn']
```

```python
for i in col:
    print(i)
    print(df[i].unique())
```

```python
for i in col:
    df[i]=df[i].apply(lambda i:0 if i=='No' else 1)
```

```python
for i in df.columns:
    if df[i].dtype=="object":
        print(i)
        print(df[i].unique())
```

```python
df["InternetService"]=df["InternetService"].map({'DSL':1, 'Fiber optic':2, 'No':0})
```

```python
df["Contract"]=df["Contract"].map({'Month-to-month':0, 'One year':1, 'Two year':2})
```

```python
df["PaymentMethod"]=df["PaymentMethod"].map({'Electronic check':1,'Mailed check':0,
                                             'Bank transfer (automatic)':2,
 'Credit card (automatic)':3})
```

```python
for i in df.columns:
    if df[i].dtype=="object":
        print(i)
        print(df[i].unique())
```

```python
df1=df[~(df.TotalCharges==" ")]
df1
```

```python
df1["TotalCharges"]=df1["TotalCharges"].apply(lambda i:float(i))
```

```python
df1.info()
```

```python
# input and output
#input
x=df1.drop("Churn",axis=1)
x
```

```python
# output
y=df1.Churn
y
```

```python
# train_test_split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print("shape of x_train=",x_train.shape)
print("shape of y_train=",y_train.shape)
print("shape of x_test=",x_test.shape)
print("shape of y_test=",y_test.shape)
```

```python
# apply random forest classifier
from sklearn.ensemble import RandomForestClassifier
model=RandomForestClassifier(n_estimators=100)
model.fit(x_train,y_train)
```

```python
# model score
model.score(x_test,y_test)*100
```

```python
# prediction using model
prediction=model.predict(x_test)
prediction[:20]
```
